```python
import pandas as pd

# Load employer dataset

df_employer = pd.read_excel(r"C:\Users\User\Desktop\youth unemployement\Employers’_Perspectives_on_Yout2026-08-31_04_33_17.xlsx")
# Basic overview
print("Shape:", df_employer.shape)

print("\nColumn names:")
for i, col in enumerate(df_employer.columns):
    print(f"{i}: {col}")

print("\nData types:")
print(df_employer.dtypes)

print("\nMissing values:")
print(df_employer.isna().sum())

print("\nUnique values per feature:")
print(df_employer.nunique(dropna=False))
```

```python
new_names_employer = [
    "submission_date",
    "industry",
    "organization_size",
    "hiring_role",
    "youth_hired_12m",
    "hiring_challenges",
    "recruitment_sources",
    "top_interview_factors",
    "reasons_not_selected",
    "importance_technical_skills",
    "importance_practical_experience",
    "importance_communication",
    "importance_problem_solving",
    "importance_ability_to_learn",
    "importance_degree",
    "importance_previous_internship",
    "importance_language_skills",
    "entry_level_job_availability",
    "expected_previous_experience",
    "response_to_skill_gap",
    "importance_connections",
    "university_preference",
    "unnecessary_requirements",
    "youth_turnover_reasons",
    "employment_improvements"
]

df_employer.columns = new_names_employer
```

```python
for i, col in enumerate(df_employer.columns):
    print(i, ":", col)
```

```python
# Remove submission date
df_employer = df_employer.drop(columns=["submission_date"])

# Create unique employer ID
df_employer.insert(
    0,
    "emp_id",
    [f"EMP{i:03d}" for i in range(1, len(df_employer) + 1)]
)

# Check
print(df_employer.shape)
print(df_employer.columns.tolist())
```

```python
#employer id display
print(df_employer["emp_id"].head())
```

```python
# Lowercase all text/categorical features
for col in df_employer.columns:
    if col != "emp_id":
        df_employer[col] = (
            df_employer[col]
            .astype("string")
            .str.lower()
            .str.strip()
        )
```

```python
# Convert all features except emp_id to categorical
for col in df_employer.columns:
    if col != "emp_id":
        df_employer[col] = df_employer[col].astype("category")

# Check data types
print(df_employer.dtypes)
```

```python
# Show all categories for each categorical feature

for col in df_employer.select_dtypes(include="category").columns:
    print("\n" + "=" * 80)
    print(col)
    print("=" * 80)
    
    for i, category in enumerate(df_employer[col].cat.categories, start=1):
        print(f"{i}. {category}")
```

```python
from pandas.api.types import CategoricalDtype

# ============================================================
# ORDERED CATEGORICAL VARIABLES
# ============================================================

# Organization size
df_employer["organization_size"] = df_employer["organization_size"].astype(
    CategoricalDtype(
        categories=[
            "1–10",
            "11–50",
            "51–250",
            "251–500",
            "more than 500"
        ],
        ordered=True
    )
)


# Importance variables
importance_levels = [
    "not important",
    "slightly important",
    "moderately important",
    "very important",
    "extremely important"
]

importance_cols = [
    "importance_technical_skills",
    "importance_practical_experience",
    "importance_communication",
    "importance_problem_solving",
    "importance_ability_to_learn",
    "importance_degree",
    "importance_previous_internship",
    "importance_language_skills"
]

for col in importance_cols:
    df_employer[col] = df_employer[col].astype(
        CategoricalDtype(
            categories=importance_levels,
            ordered=True
        )
    )


# Expected previous experience
df_employer["expected_previous_experience"] = df_employer[
    "expected_previous_experience"
].astype(
    CategoricalDtype(
        categories=[
            "no experience",
            "less than 1 year",
            "1–2 years",
            "2–3 years",
            "depends strongly on the position"
        ],
        ordered=True
    )
)


# Importance of connections
df_employer["importance_connections"] = df_employer[
    "importance_connections"
].astype(
    CategoricalDtype(
        categories=[
            "not important at all",
            "slightly important",
            "moderately important",
            "very important",
            "extremely important",
            "we mainly use open / public recruitment processes"
        ],
        ordered=True
    )
)


# Unnecessary requirements
df_employer["unnecessary_requirements"] = df_employer[
    "unnecessary_requirements"
].astype(
    CategoricalDtype(
        categories=[
            "never",
            "rarely",
            "sometimes",
            "often",
            "very often",
            "not sure",
            "depends on the position"
        ],
        ordered=True
    )
)
```

```python
# Missing values summary
missing_summary = (
    df_employer.isna()
    .sum()
    .to_frame("missing_count")
)

missing_summary["missing_percentage"] = (
    missing_summary["missing_count"] / len(df_employer) * 100
).round(1)

display(
    missing_summary[
        missing_summary["missing_count"] > 0
    ]
)
```

```python
df_employer["youth_hired_12m"] = (
    df_employer["youth_hired_12m"]
    .cat.add_categories(["none"])
    .fillna("none")
)
```

```python
import re

multi_choice_cols_employer = [
    "hiring_challenges",
    "recruitment_sources",
    "top_interview_factors",
    "reasons_not_selected",
    "youth_turnover_reasons",
    "employment_improvements"
]

# Choices to exclude from dummy creation
exclude_choices = {
    "we did not experience significant difficulty hiring",
    "we did not have positions to fill"
}

for col in multi_choice_cols_employer:

    choices = set()

    for response in df_employer[col].dropna():
        choices.update(str(response).split("\n"))

    choices = sorted(
        choice.strip()
        for choice in choices
        if choice.strip()
        and choice.strip().lower() not in exclude_choices
    )

    responses = df_employer[col].astype("string").fillna("")

    for choice in choices:

        dummy_name = (
            col + "__" +
            re.sub(r"[^a-zA-Z0-9]+", "_", choice)
            .strip("_")
            .lower()
        )

        df_employer[dummy_name] = (
            responses
            .apply(lambda x: int(choice in x.split("\n")))
            .astype("int8")
        )


# Check created dummy variables
dummy_cols_employer = [
    col for col in df_employer.columns
    if any(
        col.startswith(mc + "__")
        for mc in multi_choice_cols_employer
    )
]

print("Number of dummy variables:", len(dummy_cols_employer))

for col in dummy_cols_employer:
    print(col)
```

```python
def categorical_summary(df, col):
    summary = (
        df[col]
        .value_counts(dropna=False)
        .rename_axis("category")
        .reset_index(name="frequency")
    )

    summary["percentage"] = (
        summary["frequency"] / len(df) * 100
    ).round(1)

    return summary


profile_vars_employer = [
    "industry",
    "organization_size",
    "hiring_role",
    "youth_hired_12m"
]

for col in profile_vars_employer:
    print("\n" + "=" * 80)
    print(col)
    print("=" * 80)
    display(categorical_summary(df_employer, col))
```

```python
challenge_cols = [
    col for col in df_employer.columns
    if col.startswith("hiring_challenges__")
]

challenge_summary = pd.DataFrame({
    "challenge": [
        col.replace("hiring_challenges__", "").replace("_", " ")
        for col in challenge_cols
    ],
    "frequency": [
        df_employer[col].sum()
        for col in challenge_cols
    ]
})

challenge_summary["percentage"] = (
    challenge_summary["frequency"] / len(df_employer) * 100
).round(1)

challenge_summary = challenge_summary.sort_values(
    "frequency", ascending=False
).reset_index(drop=True)

display(challenge_summary)
```

```python
import pandas as pd

# ============================================================
# HIRING CHALLENGES — DESCRIPTIVE SUMMARY
# ============================================================

challenge_cols = [
    col for col in df_employer.columns
    if col.startswith("hiring_challenges__")
]

challenge_summary = pd.DataFrame({
    "challenge": [
        col.replace("hiring_challenges__", "").replace("_", " ")
        for col in challenge_cols
    ],
    "frequency": [
        df_employer[col].sum()
        for col in challenge_cols
    ]
})

challenge_summary["percentage"] = (
    challenge_summary["frequency"] / len(df_employer) * 100
).round(1)

challenge_summary = (
    challenge_summary
    .sort_values("frequency", ascending=False)
    .reset_index(drop=True)
)

display(challenge_summary)
```

```python
# ============================================================
# WHAT EMPLOYERS LOOK FOR
# ============================================================

def dummy_summary(df, prefix):
    
    cols = [
        col for col in df.columns
        if col.startswith(prefix + "__")
    ]

    summary = pd.DataFrame({
        "category": [
            col.replace(prefix + "__", "").replace("_", " ")
            for col in cols
        ],
        "frequency": [
            df[col].sum()
            for col in cols
        ]
    })

    summary["percentage"] = (
        summary["frequency"] / len(df) * 100
    ).round(1)

    return (
        summary
        .sort_values("frequency", ascending=False)
        .reset_index(drop=True)
    )


# ------------------------------------------------------------
# 1. Top factors employers consider when interviewing
# ------------------------------------------------------------

print("=" * 80)
print("TOP INTERVIEW FACTORS")
print("=" * 80)

top_factors_summary = dummy_summary(
    df_employer,
    "top_interview_factors"
)

display(top_factors_summary)


# ------------------------------------------------------------
# 2. Reasons young applicants are not selected
# ------------------------------------------------------------

print("=" * 80)
print("REASONS NOT SELECTED")
print("=" * 80)

reasons_summary = dummy_summary(
    df_employer,
    "reasons_not_selected"
)

display(reasons_summary)
```

```python
# ============================================================
# STEP 5 — IMPORTANCE RATINGS
# ============================================================

importance_vars = [
    "importance_technical_skills",
    "importance_practical_experience",
    "importance_communication",
    "importance_problem_solving",
    "importance_ability_to_learn",
    "importance_degree",
    "importance_previous_internship",
    "importance_language_skills"
]

# Order of the Likert scale
importance_order = [
    "not important",
    "slightly important",
    "moderately important",
    "very important",
    "extremely important"
]

# Numeric mapping used ONLY temporarily for median
importance_numeric = {
    "not important": 1,
    "slightly important": 2,
    "moderately important": 3,
    "very important": 4,
    "extremely important": 5
}

for col in importance_vars:

    print("\n" + "=" * 80)
    print(col)
    print("=" * 80)

    # --------------------------------------------------------
    # Frequency and percentage
    # --------------------------------------------------------

    summary = (
        df_employer[col]
        .value_counts(dropna=False)
        .rename_axis("category")
        .reset_index(name="frequency")
    )

    summary["percentage"] = (
        summary["frequency"] / len(df_employer) * 100
    ).round(1)

    # Order categories correctly
    summary["category"] = pd.Categorical(
        summary["category"],
        categories=importance_order,
        ordered=True
    )

    summary = (
        summary
        .sort_values("category")
        .reset_index(drop=True)
    )

    display(summary)

    # --------------------------------------------------------
    # Temporary numeric conversion ONLY for median
    # Does NOT modify df_employer
    # --------------------------------------------------------

    temp_numeric = (
        df_employer[col]
        .map(importance_numeric)
        .astype("float")
    )

    print("Median:", temp_numeric.median())
```

```python
# ============================================================
# STEP 6 — ENTRY-LEVEL HIRING CONDITIONS
# ============================================================

entry_level_vars = [
    "entry_level_job_availability",
    "expected_previous_experience",
    "unnecessary_requirements"
]

for col in entry_level_vars:

    print("\n" + "=" * 80)
    print(col)
    print("=" * 80)

    summary = (
        df_employer[col]
        .value_counts(dropna=False)
        .rename_axis("category")
        .reset_index(name="frequency")
    )

    summary["percentage"] = (
        summary["frequency"] / len(df_employer) * 100
    ).round(1)

    display(summary)
```

```python
# ============================================================
# STEP 7 — EMPLOYER RESPONSE TO SKILL GAPS & ACCESS TO JOBS
# ============================================================

step7_vars = [
    "response_to_skill_gap",
    "importance_connections",
    "university_preference"
]

for col in step7_vars:

    print("\n" + "=" * 80)
    print(col)
    print("=" * 80)

    summary = (
        df_employer[col]
        .value_counts(dropna=False)
        .rename_axis("category")
        .reset_index(name="frequency")
    )

    summary["percentage"] = (
        summary["frequency"] / len(df_employer) * 100
    ).round(1)

    display(summary)
```

```python
# ============================================================
# STEP 8 — YOUTH TURNOVER REASONS
# ============================================================

turnover_dummy_cols = [
    col for col in df_employer.columns
    if col.startswith("youth_turnover_reasons__")
]

turnover_summary = []

for col in turnover_dummy_cols:

    frequency = df_employer[col].sum()
    percentage = (frequency / len(df_employer) * 100).round(1)

    # Make the label readable
    label = col.replace("youth_turnover_reasons__", "").replace("_", " ")

    turnover_summary.append({
        "reason": label,
        "frequency": frequency,
        "percentage": percentage
    })

turnover_summary = (
    pd.DataFrame(turnover_summary)
    .sort_values("frequency", ascending=False)
    .reset_index(drop=True)
)

display(turnover_summary)
```

```python
# ============================================================
# STEP 9 — EMPLOYMENT IMPROVEMENTS
# ============================================================

improvement_dummy_cols = [
    col for col in df_employer.columns
    if col.startswith("employment_improvements__")
]

improvement_summary = []

for col in improvement_dummy_cols:

    frequency = df_employer[col].sum()
    percentage = (frequency / len(df_employer) * 100).round(1)

    # Make the label readable
    label = (
        col
        .replace("employment_improvements__", "")
        .replace("_", " ")
    )

    improvement_summary.append({
        "improvement": label,
        "frequency": frequency,
        "percentage": percentage
    })

improvement_summary = (
    pd.DataFrame(improvement_summary)
    .sort_values("frequency", ascending=False)
    .reset_index(drop=True)
)

display(improvement_summary)
```
