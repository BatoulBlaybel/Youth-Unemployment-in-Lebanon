```python
import re
import pandas as pd

# ============================================================
# MULTI-CHOICE QUESTIONS
# ============================================================
#these dummy variables are created for multi-choice questions where respondents can select multiple options. Each unique choice will be represented as a separate binary (0/1) column.

multi_choice_cols = [
    "job_opportunity_source",
    "job_search_barriers",
    "missing_skills",
    "interview_difficulties",
    "job_acceptance_barriers",
    "reasons_work_abroad",
    "employment_support"
]


# ============================================================
# SPECIAL RESPONSES TO EXCLUDE FROM REGULAR DUMMIES
# ============================================================

exclude_choices = {
    "job_opportunity_source": [
        "none of the above",
        "i rarely hear about suitable opportunities"
    ],

    "job_search_barriers": [
        "none of the above"
    ],

    "missing_skills": [
        "none of the above",
        "i do not feel unprepared"
    ],

    "interview_difficulties": [
        "none of the above",
        "i have not attended an interview",
        "i generally feel comfortable in interviews"
    ],

    "job_acceptance_barriers": [
        "none of the above"
    ],

    "reasons_work_abroad": [
        "none of the above",
        "i would prefer to remain in lebanon"
    ],

    "employment_support": [
        "none of the above"
    ]
}


# ============================================================
# CREATE DUMMY VARIABLES
# ============================================================

created_dummies = {}

for col in multi_choice_cols:

    # Get all individual choices
    choices = set()

    for response in df_youth[col].dropna():

        for choice in str(response).split("\n"):

            choice = choice.strip().lower()

            if choice:
                choices.add(choice)

    # Remove special responses
    choices = [
        choice
        for choice in choices
        if choice not in exclude_choices.get(col, [])
    ]

    question_dummies = []

    # Create 0/1 dummy for each genuine choice
    for choice in sorted(choices):

        # Create clean variable name
        clean_choice = re.sub(
            r"[^a-zA-Z0-9]+",
            "_",
            choice
        ).strip("_").lower()

        dummy_name = f"{col}__{clean_choice}"

        df_youth[dummy_name] = (
            df_youth[col]
            .fillna("")
            .astype(str)
            .apply(
                lambda x: int(
                    choice in [
                        c.strip().lower()
                        for c in x.split("\n")
                    ]
                )
            )
            .astype("int8")
        )

        question_dummies.append(dummy_name)

    created_dummies[col] = question_dummies


# ============================================================
# DISPLAY CREATED DUMMIES
# ============================================================

print("=" * 80)
print("DUMMY VARIABLES CREATED")
print("=" * 80)

for question, dummies in created_dummies.items():

    print(f"\n{question}: {len(dummies)} dummy variables")

    for dummy in dummies:
        print(f"   {dummy}")
```

```python
# ============================================================
# FREQUENCY OF EACH MULTI-CHOICE OPTION
# ============================================================

for question, dummies in created_dummies.items():

    print("\n" + "=" * 80)
    print(question)
    print("=" * 80)

    results = pd.DataFrame({
        "count": df_youth[dummies].sum(),
        "percentage": (
            df_youth[dummies].mean() * 100
        ).round(1)
    })

    results = results.sort_values(
        "count",
        ascending=False
    )

    # Make the option names easier to read
    results.index = [
        x.replace(question + "__", "").replace("_", " ")
        for x in results.index
    ]

    display(results)
```

```python

#Sample profile

def categorical_summary(df, col):
    summary = (
        df[col]
        .value_counts(dropna=False)
        .rename_axis("category")
        .reset_index(name="frequency")
    )

    summary["percentage"] = (
        summary["frequency"] / len(df) * 100
    ).round(1)

    return summary


profile_vars = [
    "education_level",
    "field_of_study",
    "current_status"
]

for col in profile_vars:
    print("\n" + "=" * 80)
    print(col)
    print("=" * 80)
    display(categorical_summary(df_youth, col))
```

```python
# ============================================================
# RANK JOB-SEARCH BARRIERS
# ============================================================

barrier_cols = {
    "Lack of practical work experience":
        "job_search_barriers__lack_of_practical_work_experience",

    "Lack of technical/digital skills":
        "job_search_barriers__lack_of_technical_or_digital_skills_required_by_employers",

    "Lack of soft skills":
        "job_search_barriers__lack_of_soft_skills_required_by_employers_e_g_communication_teamwork_adaptability",

    "Lack of professional connections/networking":
        "job_search_barriers__lack_of_professional_connections_or_networking_opportunities",

    "Limited job opportunities in my field":
        "job_search_barriers__limited_job_opportunities_in_my_field",

    "High competition":
        "job_search_barriers__high_competition_for_available_positions",

    "Location/distance":
        "job_search_barriers__location_of_available_jobs_distance_from_where_i_live",

    "Transportation difficulties":
        "job_search_barriers__transportation_or_commuting_difficulties",

    "Salary levels offered":
        "job_search_barriers__salary_levels_offered_by_employers",

    "Discrimination/bias":
        "job_search_barriers__discrimination_or_bias_during_the_hiring_process",

    "Stress/discouragement":
        "job_search_barriers__stress_discouragement_or_loss_of_motivation_during_the_job_search"
}

barrier_summary = []

for label, col in barrier_cols.items():
    count = df_youth[col].sum()
    percentage = count / len(df_youth) * 100

    barrier_summary.append({
        "barrier": label,
        "frequency": int(count),
        "percentage": round(percentage, 1)
    })

barrier_summary = (
    pd.DataFrame(barrier_summary)
    .sort_values("percentage", ascending=False)
    .reset_index(drop=True)
)

barrier_summary.index = barrier_summary.index + 1

display(barrier_summary)
```

```python
from scipy.stats import chi2_contingency
import numpy as np

table = pd.crosstab(
    df_youth["field_department"],
    df_youth["job_search_barriers__lack_of_practical_work_experience"]
)

chi2, p, dof, expected = chi2_contingency(table)

n = table.sum().sum()
r, k = table.shape

cramers_v = np.sqrt(
    chi2 / (n * min(r - 1, k - 1))
)

print("Chi-square:", round(chi2, 3))
print("Degrees of freedom:", dof)
print("p-value:", round(p, 4))
print("Cramér's V:", round(cramers_v, 3))

print("\nMinimum expected frequency:", round(expected.min(), 2))
```

```python
#no significant association between field of study and lack of practical work experience as a barrier (p > 0.05).
```

```python
from scipy.stats import mannwhitneyu

prep_mapping = {
    "not at all": 1,
    "slightly": 2,
    "moderately": 3,
    "well": 4,
    "very well": 5
}

# Temporary numeric version
prep_numeric = df_youth["education_job_preparation"].map(prep_mapping)

# Split by lack of practical work experience
group_0 = prep_numeric[
    df_youth["job_search_barriers__lack_of_practical_work_experience"] == 0
].dropna()

group_1 = prep_numeric[
    df_youth["job_search_barriers__lack_of_practical_work_experience"] == 1
].dropna()

# Mann–Whitney U test
u_stat, p_value = mannwhitneyu(
    group_0,
    group_1,
    alternative="two-sided"
)

print("Did NOT select lack of practical experience:", len(group_0))
print("Selected lack of practical experience:", len(group_1))
print("Mann–Whitney U:", u_stat)
print("p-value:", p_value)
```

```python
print("Median - Did NOT select:", group_0.median())
print("Median - Selected:", group_1.median())

print("\nDistribution:")
print(pd.DataFrame({
    "Did NOT select": group_0.value_counts().sort_index(),
    "Selected": group_1.value_counts().sort_index()
}).fillna(0))
```
